## Effect of different Activation functions

Let's see the effect of choosing among ReLU, Sigmoid and Tanh activation functions
- Keeping the architecture identical
- but changing the activation functions of the `Dense` layers
    - the Classifier `Dense` layer always uses the sigmoid (as a binary classifier must)

We see a difference in Loss and Accuracy
- note the difference in the scale of the vertical axes between plots

<img src="images/tnn_loss_and_acc.png">

## Effect of weight initialization

Training updates an initial choice of the weights via Gradient Descent. Does the initial choice matter?

The charts above used weights drawn from a **Random Normal** distribution (mean 0, unit variance; the tour uses Glorot initialization, a better-scaled random choice). Here are the identical plots with **all weights initialized to zero**:

<img src="images/tnn_loss_and_acc_zero_init.png">

The loss and accuracy of **all** the networks are notably worse than with Random Normal initialization
- the Loss sits at about $0.693 = \ln 2$, the loss of a coin flip for a binary classifier: the network is not "learning"
- the Accuracy is no better than a coin flip
