# The threads, as questions

**Why this section?** Each question was introduced where it first mattered. Here they become the columns of one table.

| Thread | The question | Introduced in |
|:--|:--|:--|
| model | is a model of the environment needed (model-based) or only samples (model-free)? | the tour, dynamic programming |
| what is learned | state values, action values, or (later) a policy? | values and Bellman, control |
| which policy generates the data | the policy being learned (on-policy) or another (off-policy)? | sampling (light mention), control (definition) |
| backup width | how many successors does one update average over: all (full) or one sample? | sampling |
| backup depth | how many steps does one update look ahead: one, $p$, or the whole episode? | sampling |
| bias of the target | is the target's average equal to the truth? | sampling |
| variance of the target | how much does the target fluctuate between samples? | sampling |

- three words used in the table
    - **backup:** one update of a value from the values of what follows it
    - **target:** the quantity the update moves the value toward (a return, or a reward plus the estimate of the next value)
    - **bootstrapping:** a target that contains our own current estimate
- *bias* compares the average of the target with the true value; *variance* is how much it fluctuates; "lower" below means lower than Monte Carlo
- one more thread is shared by the *control* rows: **evaluate versus improve** (control alternates the two); the *prediction* rows only evaluate


## Table 1: the value-based methods

**Why this section?** The whole of Part A in one grid.

| Method | Model needed? | Learns | Data from | Width | Depth | Bias of target | Variance of target |
|:--|:--|:--|:--|:--|:--|:--|:--|
| value iteration (control) | yes | $\statevalfun_*$ | none (planning) | all successors | 1 step | uses the current estimate (shrinks) | none from sampling |
| policy iteration (control) | yes | $\statevalfun_\pi$, then a better policy | none (planning) | all successors | 1 step, repeated per sweep | uses the current estimate (shrinks) | none from sampling |
| Monte Carlo (prediction) | no | $\statevalfun_\pi$ | the policy being evaluated | one whole episode | to the end (episodes must end) | none (first-visit) | high |
| TD(0) (prediction) | no | $\statevalfun_\pi$ | the policy being evaluated | one sample | 1 step | while the estimate is wrong | lower |
| $p$-step TD (prediction) | no | $\statevalfun_\pi$ | the policy being evaluated | one sample | $p$ steps | smaller as $p$ grows | higher as $p$ grows |
| SARSA (control) | no | $\actvalfun_\pi$ of its own (exploring) policy | **on-policy** | one sample | 1 step | while the estimate is wrong | lower |
| Q-learning (control) | no | $\actvalfun_*$ | **off-policy** (any behavior that keeps exploring) | one sample | 1 step | while the estimate is wrong; the max also biases it upward | lower |
| DQN (control) | no | $\actvalfun_*$ as a network | **off-policy** (replay) | a replayed mini-batch of samples | 1 step | a wrong network, plus the upward bias of the max | lower; replay breaks the correlation of consecutive data |

- "uses the current estimate" is bootstrapping; "the max also biases it upward" is **maximization bias**: the largest of several noisy estimates tends to be too large (named here, not treated further)
- *prediction* rows evaluate a given policy; *control* rows also find a better one


## Three patterns in the table

**Why this section?** The table is a map of trade-offs, not a list.

- **the width-by-depth dial**
    - dynamic programming: wide, shallow; Monte Carlo: narrow, to the end; TD: narrow and shallow, $p$-step between
    - depth moves two things together: how much of the target is our own estimate (bias) and how much is sampled (variance)
    - width decides whether there is sampling variance at all: a full backup over a known model has none
- **the bias/variance trade**
    - looking to the end of the episode gives no bias and high variance; one step gives lower variance and bias while the estimate is wrong
    - the same trade returns in Part B, for the weight on each gradient term
- **the on/off split**
    - SARSA learns about the policy it follows; Q-learning learns about the best policy while following another (one that keeps exploring)
    - with a one-step target that split is what makes DQN's replay legal; it returns in PPO as the probability ratio
- (the control rows read a policy off values, by argmax: the next slide)


## What the table does not have: a policy column

**Why this section?** The transition to Part B, seen on the table.

- every row learns or computes values; the control rows improve a policy only by an argmax over values (policy iteration keeps an explicit policy, improved this way)
- no row learns the policy's own parameters by gradient
- the limitation from the last session: the argmax needs few discrete actions, and the greedy policy it gives is deterministic (the exploring behavior policies are not, but they are not what is learned)
- Part B adds rows with a different entry under *Learns*: the policy itself, with **policy gradient** meaning gradient ascent on the policy's parameters to raise the expected return
    - and new columns, where the methods differ: the weight on each step, how far one update may move, and where the baseline comes from (the three slots)
